4. Mini-Project: Employee Payslip Generator
TechCorp HR wants a payslip tool. Data arrives from the HR form; the business rules are:

HRA (house rent allowance) = 40 percent of basic
Special allowance = 8,000 fixed
Gross = basic + HRA + special allowance
Deductions: PF = 12 percent of basic, professional tax = 200, income tax = 10 percent of gross
Net = gross minus all deductions; flag is_high_band when net crosses 50,000


```python
# Mini-project: Employee Payslip Generator (data from the HR form)
employee_name = input("Please Enter the name :")
employee_Id = input("Please Enter the ID :")
basic_salary = float(input("Please Enter the Salary :"))

#----------------><-----------------------------------------
HRA_RATE = 0.40
SPECIAL_ALLOWANCE = 8000.0
PF_RATE = 0.12
PROFESSIONAL_TAX = 200.0
INCOME_TAX_RATE = 0.10
#-------------------><---------------------------------------
hra = basic_salary * HRA_RATE
gross_salary = basic_salary + hra + SPECIAL_ALLOWANCE
pf_deduction = basic_salary + PF_RATE
income_tax = gross_salary * INCOME_TAX_RATE
total_deductions = pf_deduction + PROFESSIONAL_TAX + income_tax
net_salary = gross_salary - total_deductions

#-----------------><-------------------------------------------

print(f"Payslip For  : {employee_name} (ID: -> {employee_Id})")
print(f"Basic salary : {basic_salary:>12,.2f}")
print(f"HRA (40%)    : {hra:>12,.2f}")
print(f"Gross salary :  {gross_salary:>12,.2f}")
print(f"Total deductions: {total_deductions:>12,.2f}")
print(f"Net Salary   : {net_salary:>12,.2f}")
print(f"High band (>50k) : {net_salary > 50000}")
```
